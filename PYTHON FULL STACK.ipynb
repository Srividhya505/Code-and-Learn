# If-else

```python
a=int(input())   #Dynamic input
b=int(input())
if a>b:
    print("a is larger")
else:
    print("b is larger")
```

```python
a=10    #Static input
b=20
c=30
if (a>b and a>c):
    print("a is larger")
elif (b>a and b>c):
    print("b is larger")
else:
    print("c is larger")
```

```python
#TO CHECK EVEN POSITIVE,EVEN NEGATIVE,ODD POSITIVE,ODD NEGATIVE
a=5
if (a>=0 and a%2==0):
    print("Even Positive")
elif (a<0 and a%2==0):
    print("Even Negative")
elif (a>=0 and a%2==1):
    print("Odd Positive")
elif (a>=0 and a%2==1):
    print("odd negative")
```

# Nested If

```python
a=10     #if condition with another if condition
b=20
c=30
if a>b:
    if a>c:
        print(a)
    else:
        print(c)
else:
    if b>c:
        print(b)
    else:
        print(c)
    
```

```python
#If number is positive Then check if it is even or odd
a=6
if a>=0:
    if a%2==0:
        print("Even")
    else:
        print("Odd")
else:
    print("Negative")
```

# prime number or not

```python
n=int(input())
fc=0
for i in range(1,n+1):
    if (n%i==0):
        fc=fc+1
if (fc==2):
    print("prime number")
else:
    print("Not a prime number")
```
